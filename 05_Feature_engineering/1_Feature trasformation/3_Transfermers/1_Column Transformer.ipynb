***1_importing pandas, columntransfermer, onehotencoder, standardscaler, dataset and train-test-split***

```python
import pandas as pd 
import sklearn
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder,StandardScaler
```

```python
data= {
    "Age": [25, 30, 45, 35, 22, 50],
    "Salary": [50000, 60000, 80000, 65000, 45000, 90000],
    "City": ["Delhi", "Mumbai", "Delhi", "Bangalore", "Mumbai", "Bangalore"],
    "Gender": ["Male", "Female", "Female", "Male", "Male", "Female"],
    "Purchased": [0, 1, 1, 0, 0, 1],  
}
df=pd.DataFrame(data)
df.head()
```

***2_Seperating x and y columns and performing train test split***

```python
x=df.drop("Purchased",axis=1)
y=df["Purchased"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
print(x_test)
```

```python
print(x_train)
```

***3_Now creat columntransfermer object***

```python
ct=ColumnTransformer([
                     ('Num',StandardScaler(),['Age','Salary',]),
                     ('Cat',OneHotEncoder(),['City','Gender',])
                    ])
```

```python
x_train_transformed=ct.fit_transform(x_train)
```

```python
x_test_transformed=ct.fit(x_test)
```
