***1_importing libraries and dataset***

```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame({
    'Date': ['2026-01-15', '2026-02-20', '2026-03-10',
             '2026-04-25', '2026-05-05'],
    'Time': ['09:30:00', '14:45:00', '18:20:00',
             '10:15:00', '21:50:00'],
    'Sales': [2000, 3500, 1800, 4200, 3000]
})
df.head()
```

```python
df.dtypes
```

***2_converting the date and time columns to datetime formate***

```python
df['Date']=pd.to_datetime(df['Date'])
```

```python
df.dtypes
df.head()
```

***3_Now Extracting month, day and year from Date column***

```python
df['year']=df['Date'].dt.year
df['month']=df['Date'].dt.month
df['day']=df['Date'].dt.day
df['day of week']=df['Date'].dt.dayofweek
```

```python
df.head()
```

***4_now convert the time column to proper formate***

```python
df['Time']=pd.to_datetime(df['Time'],format='%H:%M:%S')
```

***5_Now lets hour and minute***

```python
df['Hour']=df['Time'].dt.hour
df['Minute']=df['Time'].dt.hour
```
