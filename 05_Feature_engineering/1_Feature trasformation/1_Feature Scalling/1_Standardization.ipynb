***1_import the libraries,dataset and StandardScaler()***

```python
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler
```

```python
df=pd.read_csv("featurization.csv")
df.head()
```

```python
x=df[["Age","Years_Experience","Education_Level","Monthly_Income","Hours_Per_Week","Projects_Completed","Training_Hours"]]
y=df["Performance_Score"]
```

***2_distribution of data for testing and trainnig***

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

***3_making scaler object and teansform the test and train data***

```python
scaler=StandardScaler()
```

```python
scaler.fit(x_train)
```

```python
x_train_scalled=scaler.transform(x_train)
```

```python
x_test_scalled=scaler.transform(x_test)
```
