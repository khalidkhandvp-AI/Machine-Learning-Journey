***1_importing the libraries and MinMaxscaler***

```python
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
df=pd.read_csv("featurization.csv")
df.head()
```

```python
x=df[["Age","Years_Experience","Education_Level","Monthly_Income","Hours_Per_Week","Projects_Completed","Training_Hours"]]
y=df["Performance_Score"]
```

***2_Now making object of standardscaler and doing data distribution for train and test***

```python
scaler=MinMaxScaler()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

***3_Fit the scaler on train data and transform the test and train data***

```python
scaler.fit(x_train)
```

```python
x_train_scaled=scaler.transform(x_train)
x_test_scaled=scaler.transform(x_test)
```
